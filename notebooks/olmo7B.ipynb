```python
from pathlib import Path
import shutil
import math
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
from matplotlib.colors import to_rgb
from IPython.display import display, SVG

ROOT = next(p for p in (Path.cwd(), *Path.cwd().parents)
            if (p / 'notebooks' / 'olmo7B.ipynb').is_file())
PAPER = ROOT / '6aa4d604d1c9e8e9c1117855'
TASKS = ['Safety', 'Sentiment', 'Politeness', 'Boxed', 'Answer-XX']
TITLES = ['Safety', 'Sentiment', 'Politeness', 'IF-Boxed', 'IF-Plain']
SLUGS = ['safety', 'sentiment', 'politeness', 'boxed', 'plain']
MODELS = ['Qwen3-1.7B', 'OLMo3-7B', 'OLMo3-32B', 'Qwen3-14B']
COLORS = dict(zip(MODELS, ['#237F91', '#AF7435', '#8771A6', '#3C8966']))
MODEL_LINES = dict(zip(MODELS, ['solid', (0, (4, 2)), (0, (4, 1.5, 1, 1.5)), (0, (1, 1.8))]))
LABELS = ['Qwen3\n1.7B', 'OLMo 3\n7B', 'OLMo 3\n32B', 'Qwen3\n14B']
STYLES = {
    ('Prefix-1', 'Additive'): ('o', 'full', 'Prefix-1 / Additive'),
    ('Prefix-1', 'COAST'): ('^', 'full', 'Prefix-1 / COAST'),
    ('Prefix-5', 'Additive'): ('o', 'left', 'Prefix-5 / Additive'),
    ('Prefix-5', 'COAST'): ('^', 'left', 'Prefix-5 / COAST'),
    ('Full', 'Additive'): ('o', 'none', 'Full / Additive'),
    ('Full', 'COAST'): ('^', 'none', 'Full / COAST'),
    ('Prompting', None): ('D', 'full', 'Prompting'),
    ('Unsteered', None): ('X', 'full', 'Unsteered'),
}

# Accept the old Prefix name as an alias without changing the source DATA.
def validate_plot_data(data):
    rows = []
    seen = set()
    for i, row in enumerate(data, 1):
        if len(row) != 6:
            raise ValueError(f"DATA row {i}: expected 6 fields, got {len(row)}")
        model, task, method, operator, x, y = row
        method = 'Prefix-1' if method == 'Prefix' else method
        key = (model, task, method, operator)
        if model not in MODELS or task not in TASKS or (method, operator) not in STYLES:
            raise ValueError(f"DATA row {i}: unknown model/task/method: {key}")
        if key in seen:
            raise ValueError(f"Duplicate evaluation setting: {key}")
        if not (0 <= x <= 100 and 0 <= y <= 100):
            raise ValueError(f"DATA row {i}: percentages must be between 0 and 100")
        seen.add(key)
        rows.append((*key, x, y))
    expected = {(model, task, method, operator)
                for model in MODELS for task in TASKS for method, operator in STYLES}
    missing = expected - seen
    if missing:
        raise ValueError(f"Missing {len(missing)} evaluation settings: {sorted(missing, key=str)}")
    return rows

PLOT_DATA = validate_plot_data(DATA)

# Independent task axes: edit these settings when refining an individual panel.
AXES = {
    'Safety':    {'xlim': (25, 69), 'ylim': (10, 100), 'metric': 'MMLU-Pro'},
    'Sentiment': {'xlim': (25, 65), 'ylim': (30, 100), 'metric': 'MMLU-Pro'},
    'Politeness':{'xlim': (25, 65), 'ylim': (50, 100), 'metric': 'MMLU-Pro'},
    'Boxed':     {'xlim': (40, 82), 'ylim': (40, 100), 'metric': 'MATH-500'},
    'Answer-XX': {'xlim': (40, 82), 'ylim': (0, 90), 'metric': 'MATH-500'},
}
MODEL_LABELS = dict(zip(MODELS, LABELS))

RC = {'font.family': 'serif', 'font.serif': ['STIXGeneral'],
      'font.size': 12, 'axes.labelsize': 13, 'axes.titlesize': 15,
      'axes.edgecolor': '#ADB5BD', 'axes.linewidth': 0.65,
      'pdf.fonttype': 42, 'ps.fonttype': 42}

def draw_task(ax, task, title, compact=False):
    config = AXES[task]
    task_rows = [r for r in PLOT_DATA if r[1] == task]
    # Expand the configured range if future values would otherwise be clipped.
    xlo = min(config['xlim'][0], math.floor(min(r[4] for r in task_rows) - 1))
    xhi = max(config['xlim'][1], math.ceil(max(r[4] for r in task_rows) + 1))
    ylo = min(config['ylim'][0], 10 * math.floor(min(r[5] for r in task_rows) / 10))
    yhi = max(config['ylim'][1], 10 * math.ceil(max(r[5] for r in task_rows) / 10))
    span = yhi - ylo
    ax.set_xlim(xlo, xhi)
    ax.set_ylim(ylo, yhi + 0.28 * span)
    ax.set_xticks(range(10 * math.ceil(xlo / 10), math.floor(xhi) + 1, 10))
    ax.set_yticks(range(ylo, yhi + 1, 10))
    ax.spines[['top', 'right']].set_visible(False)
    ax.set_axisbelow(True)
    # Grid only in the measured percentage range, leaving model labels clear.
    for y in ax.get_yticks():
        ax.hlines(y, xlo, xhi, color='#E6EAED', lw=0.55, zorder=0)
    for x in ax.get_xticks():
        ax.vlines(x, ylo, yhi, color='#E6EAED', lw=0.55, zorder=0)
    ax.spines['left'].set_bounds(ylo, yhi)
    ax.tick_params(length=2.5, color='#ADB5BD')
    ax.set_title(title, fontweight='semibold', pad=7)
    ax.set_xlabel(config['metric'] + ' Accuracy (%)', labelpad=5)
    ordered_models = sorted(MODELS, key=lambda m: sum(r[4] for r in task_rows if r[0] == m))
    label_x = []
    for model in ordered_models:
        values = [r[4] for r in task_rows if r[0] == model]
        label_x.append((min(values) + max(values)) / 2)
    gap = (xhi - xlo) * 0.18
    for i in range(1, len(label_x)):
        if label_x[i] - label_x[i-1] < gap:
            delta = (gap - (label_x[i] - label_x[i-1])) / 2
            label_x[i-1] -= delta
            label_x[i] += delta
    for i, (model, tx) in enumerate(zip(ordered_models, label_x)):
        label = MODEL_LABELS[model]
        ty = yhi + span * (0.18 if i % 2 == 0 else 0.10)
        color = COLORS[model]
        rows = [r for r in PLOT_DATA if r[:2] == (model, task)]
        # Join all eight settings in x order, solely to identify the model group.
        points = sorted(rows, key=lambda r: (r[4], -r[5]))
        ax.plot([r[4] for r in points], [r[5] for r in points],
                color=tuple(0.75 * v + 0.25 * 0.6 for v in to_rgb(color)),
                lw=0.95, alpha=0.85, linestyle=MODEL_LINES[model],
                solid_capstyle='round', dash_capstyle='round',
                solid_joinstyle='round', dash_joinstyle='round', zorder=2)
        for _, _, method, operator, x, y in rows:
            marker, fill_style, _ = STYLES[(method, operator)]
            ax.plot([x], [y], linestyle='none', marker=marker,
                    markersize=5.5 if compact else 7,
                    fillstyle=fill_style, markerfacecolor=color,
                    markerfacecoloralt='white', markeredgecolor=color,
                    markeredgewidth=0.85 if compact else 1.1, zorder=4)
        anchor = max(rows, key=lambda r: r[5])[4:]
        ax.annotate(label, xy=anchor, xytext=(tx, ty), ha='center', va='center',
                    fontsize=10.5 if compact else 12, color=color,
                    fontweight='semibold', linespacing=0.92,
                    arrowprops={'arrowstyle':'-', 'color':color, 'lw':0.8, 'linestyle':MODEL_LINES[model],
                                'shrinkA':2, 'shrinkB':4})

handles = [Line2D([], [], linestyle='none', marker=m, markersize=6,
                  fillstyle=fill_style, markerfacecolor='#56636E',
                  markerfacecoloralt='white', markeredgecolor='#56636E', label=label)
           for m, fill_style, label in STYLES.values()]

def save(fig, name):
    path = ROOT / 'figs' / (name + '.pdf')
    fig.savefig(path, bbox_inches='tight', pad_inches=0.045)
    shutil.copy2(path, PAPER / 'figs' / path.name)
    fig.savefig(path.with_suffix('.svg'), bbox_inches='tight', pad_inches=0.045)
    return path


# Allocate most vertical space to the dense region, retaining the entire lower range.
ZOOM_RANGES = {
    'Safety':    {'upper': (60, 99), 'lower': (10, 38), 'lower_ticks': [10, 20, 30]},
    'Sentiment': {'upper': (77, 98), 'lower': (35, 65), 'lower_ticks': [40, 50, 60]},
    'Politeness':{'upper': (76, 96), 'lower': (50, 61), 'lower_ticks': [50, 55, 60]},
    'Boxed':     {'upper': (79, 97), 'lower': (45, 60), 'lower_ticks': [45, 50, 55]},
    'Answer-XX': {'upper': (46, 83), 'lower': (5, 21), 'lower_ticks': [10, 20]},
}
# Every point must belong to exactly one segment, with room for the entire marker.
for task, ranges in ZOOM_RANGES.items():
    for row in (r for r in PLOT_DATA if r[1] == task):
        matches = [(lo, hi) for lo, hi in [ranges['lower'], ranges['upper']]
                   if lo <= row[5] <= hi]
        if len(matches) != 1:
            raise ValueError(f'Adjust ZOOM_RANGES[{task!r}] to include {row[:4]} at {row[5]}')
        lo, hi = matches[0]
        if min(row[5] - lo, hi - row[5]) < (hi - lo) * 0.025:
            raise ValueError(f'Increase axis padding for {row[:4]} at {row[5]}')

def draw_zoom_task(fig, slot, task, title, compact=False):
    split = slot.subgridspec(2, 1, height_ratios=[4, 1], hspace=0.08)
    top = fig.add_subplot(split[0])
    bottom = fig.add_subplot(split[1], sharex=top)
    draw_task(top, task, title, compact)
    draw_task(bottom, task, '', compact)
    config = ZOOM_RANGES[task]
    start, high = config['upper']
    low, lower_high = config['lower']
    for i, txt in enumerate(top.texts):
        tx, _ = txt.get_position()
        txt.set_position((tx, high + (0.18 if i % 2 == 0 else 0.10) * (high - start)))
    for txt in list(bottom.texts):
        txt.remove()
    for ax in (top, bottom):
        # Replace the full-range grid built by draw_task.
        for collection in list(ax.collections):
            collection.remove()
    top.set_ylim(start, high + 0.28 * (high - start))
    top.set_yticks(range(5 * math.ceil(start / 5), high + 1, 5))
    top.set_xlabel('')
    top.tick_params(axis='x', bottom=False, labelbottom=False)
    top.spines['bottom'].set_visible(False)
    top.spines['left'].set_bounds(start, high)
    bottom.set_ylim(low, lower_high)
    bottom.set_yticks(config['lower_ticks'])
    bottom.spines['left'].set_bounds(low, lower_high)
    bottom.set_title('')
    bottom.tick_params(axis='y', labelsize=10 if compact else 11)
    for ax, floor, ceiling in [(top, start, high), (bottom, low, lower_high)]:
        xlo, xhi = ax.get_xlim()
        for y in ax.get_yticks():
            ax.hlines(y, xlo, xhi, color='#E6EAED', lw=0.55, zorder=0)
        for x in ax.get_xticks():
            ax.vlines(x, floor, ceiling, color='#E6EAED', lw=0.55, zorder=0)
    # Remove off-segment point artists so clipped symbols do not appear at the seam.
    for ax, floor, ceiling in [(top, start, high), (bottom, low, lower_high)]:
        for line in list(ax.lines):
            if len(line.get_ydata()) == 1:
                y = line.get_ydata()[0]
                if y < floor or y > ceiling or (ax is bottom and y == lower_high):
                    line.remove()
    # Diagonal marks signal the change of vertical scale; only empty intervals are omitted.
    for ax, y in [(top, 0), (bottom, 1)]:
        ax.plot([-0.016, 0.016], [y-0.014, y+0.014], transform=ax.transAxes,
                color='#74818B', lw=0.9, clip_on=False)
    return top, bottom

with plt.rc_context(RC):
    for task, title, slug in zip(TASKS, TITLES, SLUGS):
        fig = plt.figure(figsize=(6.6, 4.9))
        slot = fig.add_gridspec(1, 1, left=0.11, right=0.98, bottom=0.15, top=0.74)[0]
        draw_zoom_task(fig, slot, task, title)
        fig.supylabel('Target Success (%)', x=0.015, fontsize=13)
        fig.legend(handles=handles, loc='upper center', ncol=4, frameon=False,
                   fontsize=9.5, columnspacing=1.0, handletextpad=0.4)
        save(fig, 'models_' + slug)
        plt.close(fig)

    fig = plt.figure(figsize=(10.6, 7.1))
    grid = fig.add_gridspec(2, 6, left=0.065, right=0.99, bottom=0.10,
                           top=0.83, hspace=0.50, wspace=0.75)
    positions = [(0,0,2), (0,2,4), (0,4,6), (1,1,3), (1,3,5)]
    axes = []
    for i, (task, title, (row, start, stop)) in enumerate(zip(TASKS, TITLES, positions)):
        pair = draw_zoom_task(fig, grid[row,start:stop], task, f'({chr(97+i)}) {title}', compact=True)
        axes.append(pair)
    fig.supylabel('Target Success (%)', x=0.008, fontsize=14)
    fig.legend(handles=handles, loc='upper center', bbox_to_anchor=(0.53, 1.0),
               ncol=4, frameon=False, fontsize=12, columnspacing=1.7,
               handletextpad=0.45, labelspacing=0.6)
    output = save(fig, 'models_tasks_overview')
    plt.close(fig)
    display(SVG(filename=str(output.with_suffix('.svg'))))
print('Exported five task figures and the overview: 160 points, four models, five tasks.')
```
